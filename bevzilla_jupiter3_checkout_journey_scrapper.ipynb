```python
#3
import smtplib
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart

# Define email credentials
SENDER_EMAIL = "arnav.gupta@oneguardian.in"
RECEIVER_EMAIL = "arnav12052004@gmail.com"
EMAIL_PASSWORD = ""  # ⚠️ Use environment variables in production.

# Global variable to track last error message sent
LAST_ERROR_MESSAGE = None

def send_error_email(step, exception):
    """Send an email with only the error step, without traceback."""
    global LAST_ERROR_MESSAGE

    error_message = str(exception)
    if error_message == LAST_ERROR_MESSAGE:
        return  # Avoid sending duplicate emails

    LAST_ERROR_MESSAGE = error_message  # Update last error message

    try:
        # Setup email server
        server = smtplib.SMTP("smtp.gmail.com", 587)
        server.starttls()  # Secure connection
        server.login(SENDER_EMAIL, EMAIL_PASSWORD)

        # Email content setup (Without traceback)
        subject = f"Automation Script Error at Step: {step}"
        body = f"""An error occurred in the script.

Step: {step}

Error Message: {error_message}
"""

        # Create and send the email
        message = MIMEMultipart()
        message["From"] = SENDER_EMAIL
        message["To"] = RECEIVER_EMAIL
        message["Subject"] = subject
        message.attach(MIMEText(body, "plain"))

        server.sendmail(SENDER_EMAIL, RECEIVER_EMAIL, message.as_string())
        print(f"Error email sent for step '{step}' (without traceback)")
    except Exception as email_exception:
        print(f"Failed to send error email: {email_exception}")
    finally:
        server.quit()  # Ensure the server connection is closed








# 2
# import smtplib
# import traceback
# from email.mime.text import MIMEText
# from email.mime.multipart import MIMEMultipart

# # Define email credentials
# SENDER_EMAIL = "arnav.gupta@oneguardian.in"
# RECEIVER_EMAIL = "arnav12052004@gmail.com"
# EMAIL_PASSWORD = "oiog ktli zxei soib"  # ⚠️ Use environment variables in production.

# # Global variable to track last error message sent
# LAST_ERROR_MESSAGE = None


# def send_error_email(step, exception):
#     """Send an email with the error details, ensuring no duplicate error emails are sent."""
#     global LAST_ERROR_MESSAGE

#     error_message = str(exception)
#     if error_message == LAST_ERROR_MESSAGE:
#         return  # Avoid sending duplicate emails

#     LAST_ERROR_MESSAGE = error_message  # Update last error message

#     try:
#         # Setup email server
#         server = smtplib.SMTP("smtp.gmail.com", 587)
#         server.starttls()  # Secure connection
#         server.login(SENDER_EMAIL, EMAIL_PASSWORD)

#         # Email content setup
#         subject = f"Automation Script Error at Step: {step}"
#         body = f"""An error occurred during the execution of the script.

# Step: {step}

# Error Message: {error_message}

# Traceback:
# {traceback.format_exc()}
# """

#         # Create and send the email
#         message = MIMEMultipart()
#         message["From"] = SENDER_EMAIL
#         message["To"] = RECEIVER_EMAIL
#         message["Subject"] = subject
#         message.attach(MIMEText(body, "plain"))

#         server.sendmail(SENDER_EMAIL, RECEIVER_EMAIL, message.as_string())
#         print(f"Error email sent for step '{step}'")
#     except Exception as email_exception:
#         print(f"Failed to send error email: {email_exception}")
#     finally:
#         server.quit()  # Ensure the server connection is closed







# import smtplib
# import traceback
# from email.mime.text import MIMEText
# from email.mime.multipart import MIMEMultipart

# # Email function to send exception details
# def send_error_email(step, exception):
#     # Define email credentials
#     sender = "arnav.gupta@oneguardian.in"
#     # receiver = "anshika.vishnoi@oneguardian.in"
#     receiver = "arnav12052004@gmail.com"
    
#     password = "oiog ktli zxei soib"  # ⚠️ Use app-specific password or environment variable in production for better security.

#     try:
#         # Setup email server
#         server = smtplib.SMTP("smtp.gmail.com", 587)
#         server.starttls()  # Secure connection

#         # Login to the email server
#         server.login(sender, password)

#         # Email content setup
#         subject = f"Automation Script Error at Step: {step}"
#         body = f"""An error occurred during the execution of the script.
        
# Step: {step}
# Error Message: {str(exception)}

# Traceback:
# {traceback.format_exc()}
# """

#         # Create a MIME email
#         message = MIMEMultipart()
#         message["From"] = sender
#         message["To"] = receiver
#         message["Subject"] = subject
#         message.attach(MIMEText(body, "plain"))

#         # Send the email
#         server.sendmail(sender, receiver, message.as_string())
#         print(f"Error email sent for step '{step}'")
#     except Exception as email_exception:
#         print(f"Failed to send error email: {email_exception}")
#     finally:
#         server.quit()  # Ensure the server connection is closed

# Example usage
# if __name__ == "__main__":
#     try:
#         # Simulate a step that throws an exception
#         step = "Step 1: Open website"
#         raise ValueError("Test error for demonstration purposes.")
#     except Exception as e:
        # send_error_email(step, e)
```
